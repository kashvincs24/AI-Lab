```python
board = [' '] * 9

def print_board():
    print(f"\n {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} \n")

def is_winner(b, p):
    wins = [(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    return any(b[x] == b[y] == b[z] == p for x, y, z in wins)

def minimax(b, is_max):
    if is_winner(b, 'X'): return 10
    if is_winner(b, 'O'): return -10
    if ' ' not in b: return 0

    scores = []
    for i in range(9):
        if b[i] == ' ':
            b[i] = 'X' if is_max else 'O'
            scores.append(minimax(b, not is_max))
            b[i] = ' '
    return max(scores) if is_max else min(scores)

def best_move():
    best_score, move = -100, -1
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'X'
            score = minimax(board, False)
            board[i] = ' '
            if score > best_score:
                best_score, move = score, i
    return move

# Main Game Loop
print("Positions are 1 to 9:")
print(" 1 | 2 | 3 \n---|---|---\n 4 | 5 | 6 \n---|---|---\n 7 | 8 | 9 ")

turn = input("Do you want to play first? (y/n): ").strip().lower()
human_turn = True if turn == 'y' else False

for _ in range(9):
    print_board()
    if is_winner(board, 'X'):
        print("🤖 AI Wins!")
        break
    if is_winner(board, 'O'):
        print("🎉 You Win!")
        break

    if human_turn:
        pos = int(input("Enter position (1-9): ")) - 1
        if board[pos] != ' ':
            print("Spot taken! Try again.")
            continue
        board[pos] = 'O'
    else:
        print("AI is thinking...")
        board[best_move()] = 'X'

    human_turn = not human_turn
else:
    print_board()
    print("🤝 It's a draw!")
```
