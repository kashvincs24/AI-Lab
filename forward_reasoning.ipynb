```python
# Initial Facts
facts = {
    "American(Robert)",
    "Enemy(A,America)",
    "Missile(T1)",
    "Owns(A,T1)"
}
changed = True
while changed:
    changed = False
    # Rule 1: Enemy(x,America) -> Hostile(x)
    if "Enemy(A,America)" in facts and "Hostile(A)" not in facts:
        facts.add("Hostile(A)")
        changed = True
    # Rule 2: Missile(x) -> Weapon(x)
    if "Missile(T1)" in facts and "Weapon(T1)" not in facts:
        facts.add("Weapon(T1)")
        changed = True
    # Rule 3: Missile(x) and Owns(A,x) -> Sells(Robert,x,A)
    if "Missile(T1)" in facts and "Owns(A,T1)" in facts:
        if "Sells(Robert,T1,A)" not in facts:
            facts.add("Sells(Robert,T1,A)")
            changed = True
    # Rule 4:
    # American(p) and Weapon(q) and Sells(p,q,r) and Hostile(r)
    # -> Criminal(p)
    if ("American(Robert)" in facts and
        "Weapon(T1)" in facts and
        "Sells(Robert,T1,A)" in facts and
        "Hostile(A)" in facts):
        if "Criminal(Robert)" not in facts:
            facts.add("Criminal(Robert)")
            changed = True
# Display all derived facts
print("Derived Facts:")
for fact in facts:
    print(fact)
# Query
if "Criminal(Robert)" in facts:
    print("\nQuery Proved: Criminal(Robert)")
else:
    print("\nQuery Not Proved")
```
