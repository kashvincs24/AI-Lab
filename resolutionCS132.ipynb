```python
def resolve(c1, c2):
    for literal in c1:
        opposite = "~" + literal if not literal.startswith("~") else literal[1:]

        if opposite in c2:
            result = (set(c1) - {literal}) | (set(c2) - {opposite})
            return list(result)

    return None


def resolution(KB, goal):
    KB.append([ "~" + goal ])

    while True:
        new_clauses = []

        for i in range(len(KB)):
            for j in range(i + 1, len(KB)):
                result = resolve(KB[i], KB[j])

                if result == []:
                    print("Goal Proved")
                    return True

                if result is not None and result not in KB:
                    new_clauses.append(result)

        if not new_clauses:
            print("Goal Not Proved")
            return False

        KB.extend(new_clauses)


KB = [
    ["H", "C"],
    ["~C", "R"],
    ["~H", "T"],
    ["~T"]
]

goal = "R"

resolution(KB, goal)
```
